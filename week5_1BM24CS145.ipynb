```python
import random
import math

def calculate_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == j - i:
                conflicts += 1
    return conflicts

def get_random_neighbor(state):
    neighbor = list(state)
    col = random.randint(0, 7)
    row = random.randint(0, 7)
    while row == state[col]:
        row = random.randint(0, 7)
    neighbor[col] = row
    return neighbor

def simulated_annealing():
    current_state = [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_conflicts(current_state)

    T = 1.0
    alpha = 0.99
    min_T = 0.00001

    while T > min_T and current_cost > 0:
        neighbor = get_random_neighbor(current_state)
        neighbor_cost = calculate_conflicts(neighbor)
        d_energy = neighbor_cost - current_cost

        if d_energy < 0:
            current_state = neighbor
            current_cost = neighbor_cost
        else:
            p = math.exp(-d_energy / T)
            if random.random() < p:
                current_state = neighbor
                current_cost = neighbor_cost

        T *= alpha

    return current_state, current_cost

def print_board(state):
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += "Q "
            else:
                line += ". "
        print(line.strip())

solution, final_cost = simulated_annealing()
print(f"Final State: {solution} with {final_cost} conflicts.\n")
if final_cost == 0:
    print_board(solution)
else:
    print("Could not find a perfect solution within the temperature schedule. Run again.")
```
