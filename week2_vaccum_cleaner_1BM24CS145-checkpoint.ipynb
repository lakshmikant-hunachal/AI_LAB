```python
def vacuum_cleaner():
    room = {
        "A": "Dirty",
        "B": "Dirty"
    }

    position = "A"

    while "Dirty" in room.values():
        print("Current position:", position)
        print("Room status:", room)

        if room[position] == "Dirty":
            print("Action: Suck")
            room[position] = "Clean"

        elif position == "A":
            print("Action: Move Right")
            position = "B"

        elif position == "B":
            print("Action: Move Left")
            position = "A"

    print("\nAll rooms are clean!")
    print("Final room status:", room)


vacuum_cleaner()
```

```python
#output:
# Current position: A
# Room status: {'A': 'Dirty', 'B': 'Dirty'}
# Action: Suck
# Current position: A
# Room status: {'A': 'Clean', 'B': 'Dirty'}
# Action: Move Right
# Current position: B
# Room status: {'A': 'Clean', 'B': 'Dirty'}
# Action: Suck

# All rooms are clean!
# Final room status: {'A': 'Clean', 'B': 'Clean'}
```
