```python
import random

def evaluate(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def hill_climbing(n):
    current_state = [random.randint(0, n - 1) for _ in range(n)]
    current_conflicts = evaluate(current_state)

    while current_conflicts > 0:
        neighbors = []
        neighbor_conflicts = []

        for col in range(n):
            for row in range(n):
                if row != current_state[col]:
                    temp_state = list(current_state)
                    temp_state[col] = row
                    neighbors.append(temp_state)
                    neighbor_conflicts.append(evaluate(temp_state))

        min_conflicts = min(neighbor_conflicts)
        if min_conflicts >= current_conflicts:
            current_state = [random.randint(0, n - 1) for _ in range(n)]
            current_conflicts = evaluate(current_state)
        else:
            best_indices = [i for i, c in enumerate(neighbor_conflicts) if c == min_conflicts]
            chosen_index = random.choice(best_indices)
            current_state = neighbors[chosen_index]
            current_conflicts = min_conflicts

    return current_state

def main():
    n = 4
    solution = hill_climbing(n)
    print("Solution state (column: row):", solution)

if __name__ == "__main__":
    main()
```
