```python
def print_puzzle(state):
    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])
    print()


def dfs(state, goal, depth,limit,visited, path):

    # Goal reached
    if state == goal:
        path.append(state)
        return True


    if depth==limit:
      return False
    # Mark state as visited
    visited.add(state)
    path.append(state)

    # Find blank space
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check valid position
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            # Create new state
            new_state = list(state)

            # Swap blank with tile
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            new_state = tuple(new_state)

            # DFS on new state
            if new_state not in visited:

                if dfs(new_state, goal,depth+1,limit, visited, path):
                    return True

    # Backtracking
    path.pop()

    return False


# Starting state
start = (
    2,8,3,
    1,6,4,
    7,0,5
)

# Goal state
goal = (
    1, 2, 3,
    8,0,4,
    7,6,5
)

# Visited states
visited = set()

# Solution path
path = []
limit=6
depth=0
# Run DFS
if dfs(start, goal,depth,limit, visited, path):

    print("Solution found!")
    print()

    for i, state in enumerate(path):

        print("Step", i)
        print_puzzle(state)

else:
    print("No solution found")
```
