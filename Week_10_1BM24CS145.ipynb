```python
import math


def alpha_beta_search(tree, root, is_max_root=True):
    """Implements ALPHA-BETA-SEARCH(state) from the pseudocode."""
    if is_max_root:
        best_val = -math.inf
        best_action = None
        alpha, beta = -math.inf, math.inf

        for action, child in tree[root].items():
            v = min_value(tree, child, alpha, beta)
            if v > best_val:
                best_val = v
                best_action = action
            alpha = max(alpha, best_val)
        return best_action, best_val
    else:
        best_val = math.inf
        best_action = None
        alpha, beta = -math.inf, math.inf

        for action, child in tree[root].items():
            v = max_value(tree, child, alpha, beta)
            if v < best_val:
                best_val = v
                best_action = action
            beta = min(beta, best_val)
        return best_action, best_val


def max_value(tree, state, alpha, beta):
    """Implements MAX-VALUE(state, alpha, beta)."""
    # TERMINAL-TEST(state) -> if state is a leaf (number)
    if isinstance(state, (int, float)):
        return state

    v = -math.inf
    for action, child in tree[state].items():
        v = max(v, min_value(tree, child, alpha, beta))
        if v >= beta:
            print(f"Pruned remaining branches at MAX node '{state}' (v={v} >= beta={beta})")
            return v
        alpha = max(alpha, v)
    return v


def min_value(tree, state, alpha, beta):
    """Implements MIN-VALUE(state, alpha, beta)."""
    # TERMINAL-TEST(state) -> if state is a leaf (number)
    if isinstance(state, (int, float)):
        return state

    v = math.inf
    for action, child in tree[state].items():
        v = min(v, max_value(tree, child, alpha, beta))
        if v <= alpha:
            print(f"Pruned remaining branches at MIN node '{state}' (v={v} <= alpha={alpha})")
            return v
        beta = min(beta, v)
    return v


# --- User Input Section ---
def build_tree_from_input():
    tree = {}
    n = int(input("Enter number of internal (non-terminal) nodes: "))
    print("\nFor each internal node, enter its children.")
    print("Format: NodeName : child1 child2 ... (use numbers for terminal leaves)")
    print("Example: D : 2 3\n")

    for _ in range(n):
        line = input("Node definition: ").strip()
        node, children_str = line.split(":")
        node = node.strip()
        children = children_str.strip().split()

        tree[node] = {}
        for i, child in enumerate(children, start=1):
            # Convert numeric strings to int/float for terminal nodes
            try:
                val = int(child)
            except ValueError:
                try:
                    val = float(child)
                except ValueError:
                    val = child
            action_name = f"{node}->{child}" if isinstance(val, str) else f"{node}->Leaf({val})"
            tree[node][action_name] = val

    root = input("\nEnter root node name: ").strip()
    root_type = input("Is root a MAX node? (y/n): ").strip().lower()
    return tree, root, (root_type == 'y')


if __name__ == "__main__":
    game_tree, root_node, is_max = build_tree_from_input()
    print("\n--- Running Alpha-Beta Search ---")
    optimal_action, optimal_value = alpha_beta_search(game_tree, root_node, is_max)
    print(f"\nOptimal Value at Root ({root_node}): {optimal_value}")
    print(f"Best First Move: {optimal_action}")
```
